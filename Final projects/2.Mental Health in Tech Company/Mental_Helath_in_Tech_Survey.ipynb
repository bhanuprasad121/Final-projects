## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# copy the data
df1=df.copy()

# checking missing values
print("Checking missing values",df1.isnull().sum())
print()
df1["state"].fillna("Unkown",inplace=True)
df1["self_employed"].fillna("No",inplace=True)
df1["work_interfere"].fillna("Don't know",inplace=True)
print("After cleaning the missing values")
print("Checking missing values",df1.isnull().sum())
print()
# cleaning messy data
df1=df1[(df1["Age"]>=18) &(df1["Age"]<=100)]
df1=df1.drop(["Timestamp","comments"],axis=1)
df1["Gender"]=df1["Gender"].str.lower()
male = ['male','m','man','cis male','male-ish','msle','mal','mail','make']
female = ['female','f','woman','cis female','femake','female ','cis-female']
def gender_cleaning(x):
  if x in male:
    return "Male"
  elif x in female:
    return "Female"
  else:
    return "Other"

df1["Gender"] = df1["Gender"].apply(gender_cleaning)

df1["no_employees"] = df1["no_employees"].replace({
    "Jun-25": "6-25",
    "01-May": "1-5"
})

print(df1.columns)

# duplicate values
print("count of duplicate values after cleaning",df1.duplicated().sum())
print("Duplicate values")
print("count of duplicate values",df1.duplicated().sum())
df1=df1.drop_duplicates()
print()
print(df1.info())
````

### What all manipulations have you done and insights you found?

The dataset was cleaned and prepared for analysis by first identifying missing values across all columns. Missing values in the state, self_employed, and work_interfere columns were handled by replacing them with "Unknown", "No", and "Don't know", respectively. Records with unrealistic age values were removed by retaining only respondents between 18 and 100 years. The Timestamp and comments columns were dropped as they were not relevant to the analysis. The Gender column was standardized by converting all values to lowercase to ensure consistency. Additionally, duplicate records were identified and removed, resulting in a clean dataset containing 1,247 records and 25 columns. After these preprocessing steps, all remaining columns were free from missing values (except the removed comments column), making the dataset reliable and ready for exploratory data analysis and visualization.

#### Chart - 6 Countries with work Interference

````python
# Chart - 6 Countries Wiith Interference
country_total = df1.groupby("Country").size()

valid_countries = country_total[country_total >= 20].index

country_work = (
    df1[df1["work_interfere"].isin(["Often", "Sometimes"])]
    .groupby("Country")
    .size()
    .div(country_total[valid_countries])
    .mul(100)
).sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
ax=sns.barplot(
    x=country_work.values,
    y=country_work.index,
    palette="viridis"
)

for i, v in enumerate(country_work.values):
    ax.text(v + 0.5, i, f"{v:.1f}%", va="center", fontsize=10)

plt.title("Countries with Work Interference")
plt.xlabel("Percentage")
plt.ylabel("Country")
plt.show()
````

#### Chart - 9 Willingness To Discuss Mental Health

````python
# Chart - 9 Willingness To Discuss Mental Health
coworkers = df1["coworkers"].value_counts(normalize=True) * 100
supervisor = df1["supervisor"].value_counts(normalize=True) * 100
interview = df1["mental_health_interview"].value_counts(normalize=True) * 100

discussion = pd.DataFrame({
    "Coworkers": coworkers,
    "Supervisors": supervisor,
    "Interview": interview
}).fillna(0)

discussion = discussion.T

plt.figure(figsize=(9,6))
ax = discussion.plot(
    kind="bar",
    figsize=(9,6),
    colormap="viridis"
)

plt.title("Willingness to Discuss Mental Health")
plt.xlabel("Discussion With")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.legend(title="Response")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", fontsize=9)

plt.show()
````

##### 1. Why did you pick the specific chart?

A grouped bar chart was selected to compare respondents' willingness to discuss mental health in three different situations: with coworkers, supervisors, and during job interviews. ##### 2. What is/are the insight(s) found from the chart?

Supervisors: Employees are more comfortable discussing mental health with supervisors than expected, with 41.1% answering Yes, although 31.1% still prefer not to discuss it.
Job Interviews: Respondents are least willing to discuss mental health during interviews. A significant 80.2% answered No, while only 3.3% said Yes, indicating a strong fear of stigma or negative career impact.

#### Chart - 14 - Correlation Heatmap

````python
# encodeing categorical values
from sklearn.preprocessing import LabelEncoder

df_corr = df1.copy()

le = LabelEncoder()

for col in df_corr.select_dtypes(include="object").columns:
    df_corr[col] = le.fit_transform(df_corr[col].astype(str))

corr_matrix = df_corr.corr()
````

````python
# Correlation Heatmap visualization code
plt.figure(figsize=(16,12))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Mental Health Survey")
plt.show()
````

##### 1. Why did you pick the specific chart?

##### 2. What is/are the insight(s) found from the chart?


The strongest positive correlation is between Treatment and Work Interference (0.62), indicating that employees whose mental health affects their work are more likely to seek treatment.
Coworkers and Supervisors show a strong positive correlation (0.57), suggesting that employees who feel comfortable discussing mental health with coworkers are also likely to feel comfortable discussing it with supervisors.
Wellness Program and Seek Help have a moderate positive correlation (0.47), indicating that organizations offering wellness programs are more likely to encourage employees to seek mental health support.
Benefits and Care Options are moderately correlated (0.44), suggesting that companies providing mental health benefits often also provide professional care options.
Family History has a moderate positive relationship with Treatment (0.37), implying that employees with a family history of mental illness are more likely to receive treatment.
Most other variables have weak correlations (close to 0), indicating limited direct linear relationships.
