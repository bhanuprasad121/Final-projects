## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df1_Copy=df1.copy()
df2_Copy=df2.copy()

# checking missing values
print("Missing Values in Title Dataset")
display(df1_Copy.isnull().sum())
print("\n")
print("Missing Values in Credit Dataset")
display(df2_Copy.isnull().sum())
print()
print(df1_Copy.info())
print()
print(df1_Copy.describe(include="all"))

#Checking Duplicate Values
print("Before Deleting Duplicates")
print("Duplicate Values in Title Dataset:",df1_Copy.duplicated().sum())
print("/n")
print("Duplicate Values in Credit Dataset:",df2_Copy.duplicated().sum())
# Deleting Duplicates
print("After Deleting Duplicates")
df1_Copy.drop_duplicates(inplace=True)
df2_Copy.drop_duplicates(inplace=True)
print("Duplicate Values in Title Dataset:",df1_Copy.duplicated().sum())
print("/n")
print("Duplicate Values in Credit Dataset:",df2_Copy.duplicated().sum())

# Cleaning
df1_Copy["type"]=df1_Copy["type"].str.lower()
df1_Copy["genres"]=df1_Copy["genres"].str.lower()
df1_Copy["production_countries"]=df1_Copy["production_countries"].str.lower()
df1_Copy["genres"]=df1_Copy["genres"].str.replace(r"[\[\]']", "", regex=True)
df1_Copy["production_countries"]=df1_Copy["production_countries"].str.replace(r"[\[\]']", "", regex=True)
````

#### Chart - 5  Country-wise Analysis

````python
country_df = df1_Copy.copy()
import ast
def clean_country(value):

    if isinstance(value, list):
        return value

    if pd.isna(value):
        return []

    value = str(value).strip()

    try:
        result = ast.literal_eval(value)
        if isinstance(result, list):
            return result
    except (ValueError, SyntaxError):
        pass

    return [item.strip() for item in value.split(",") if item.strip()]

country_df["production_countries"] = (
    country_df["production_countries"].apply(clean_country)
)

country_df = country_df.explode("production_countries")

country_df["production_countries"] = (
    country_df["production_countries"].astype(str).str.strip()
)

country_df = country_df[
    country_df["production_countries"].notna()
    & ~country_df["production_countries"].isin(["", "[]", "nan", "None"])
]

country_count = (
    country_df["production_countries"]
    .value_counts()
    .head(10)
)

print(country_count)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=country_count.values,
    y=country_count.index
)

if ax.containers:
    ax.bar_label(ax.containers[0], padding=3)

plt.title("Top 10 Production Countries")
plt.xlabel("Number of Titles")
plt.ylabel("Production Country")
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A horizontal bar chart was chosen because it clearly compares the number of titles produced by different countries. It is ideal for ranking countries from highest to lowest production and is easier to read than a vertical bar chart when country names are displayed on the y-axis.

##### 2. What is/are the insight(s) found from the chart?

The United States (US) is the largest content producer on Amazon Prime Video with 5,331 titles, contributing significantly more content than any other country.

#### Chart - 6 Year Vs Genres

````python
# Chart - 6 visualization code
heatmap_df = df1_Copy.copy()

heatmap_df["genres"] = heatmap_df["genres"].str.split(",")

heatmap_df = heatmap_df.explode("genres")

heatmap_df["genres"] = heatmap_df["genres"].str.strip()

top_genres = heatmap_df["genres"].value_counts().head(10).index
heatmap_df = heatmap_df[heatmap_df["genres"].isin(top_genres)]

pivot = pd.pivot_table(
    heatmap_df,
    values="id",
    index="genres",
    columns="release_year",
    aggfunc="count",
    fill_value=0
)

plt.figure(figsize=(16,6))

sns.heatmap(
    pivot,
    cmap="viridis",
    linewidths=0.5,
    linecolor="white"
)

plt.title("Year Vs Genres")
plt.xlabel("Release Year")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
````
