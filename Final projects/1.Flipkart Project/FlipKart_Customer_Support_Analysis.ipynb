## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# makeing copy of original dataset
df1=df.copy()

# checking missing values
print("Checking Missing Values")
print(df1.isnull().sum())
print()

# Drop duplicate values
df1.drop_duplicates(inplace=True)
print("Droped Duplicates")
print()

# covert date columns to time date columns
date_cols=["order_date_time",
    "Issue_reported at",
    "issue_responded",
    "Survey_response_Date"]
for col in date_cols:
  df[col]=pd.to_datetime(df[col],errors="coerce")
  print(f"Converted {col} to datetime")
print()

# filling missing values
df1["Customer Remarks"].fillna("Not provided",inplace=True)
df1["Order_id"].fillna("Unknown",inplace=True)
df1["Customer_City"].fillna("Unknown",inplace=True)
df1["Product_category"].fillna("Unknown",inplace=True)
df1["Item_price"].fillna(df1["Item_price"].median(),inplace=True)

# data types
print("Checking Data types")
print(df1.dtypes)
print()

print(df1.info())
````

### What all manipulations have you done and insights you found?

The dataset was prepared for analysis by first checking for duplicate records and missing values. Duplicate rows were removed to maintain data quality. The date columns (order_date_time, Issue_reported at, issue_responded, and Survey_response_Date) were converted into the datetime format for easier time-based analysis. Missing values in the Item_price column were replaced with the median value to preserve the dataset while minimizing the effect of outliers. The connected_handling_time column contained a very high number of missing values, so it was excluded from further analysis. After these preprocessing steps, the dataset became clean, consistent, and ready for exploratory data analysis. These manipulations ensured more reliable insights and improved the overall quality of the analysis.
