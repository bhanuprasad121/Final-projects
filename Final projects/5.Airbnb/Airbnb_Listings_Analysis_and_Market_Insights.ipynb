## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Create a copy of the original DataFrame
df1 = df.copy()

print("Checking Duplicate value")
print(df1.duplicated().sum())
print()
print("Droping Unnecessary columns")
df1.columns
df1=df1.drop(["neighbourhood_group","minimum_nights","number_of_reviews","last_review","reviews_per_month","calculated_host_listings_count","availability_365","number_of_reviews_ltm","license"],axis=1)
print("Droped Columns")
print(df1.columns)
print()
print("Checking null value")
print(df1.isnull().sum())
df1["price"].fillna(df1["price"].mean(),inplace=True)
print("Null values have removed")
print(df1.isnull().sum())
````

### What all manipulations have you done and insights you found?

The dataset was cleaned and prepared for analysis by performing several data preprocessing steps. First, the dataset was checked for duplicate records, and no duplicate values were found. Next, unnecessary columns that were not required for the analysis were removed, reducing the dataset to the most relevant features such as listing ID, property name, host details, neighborhood, location coordinates, room type, and price. The dataset was then examined for missing values, where the price column was found to contain 32 null values. These missing values were handled to ensure the dataset contained no null values before analysis. After completing the cleaning process, the dataset became consistent, accurate, and ready for further exploratory data analysis and visualization.

From the cleaned dataset, it was observed that there were no duplicate listings, indicating good data quality. The only significant missing values were found in the price column, while the remaining selected columns were complete. The cleaned data can now be used to analyze pricing patterns across neighborhoods, compare room types, study host listings, and understand the geographical distribution of Airbnb properties, enabling meaningful business insights and better decision-making.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1 Price Distribution

````python
# Chart - 1 visualization code
plt.figure(figsize=(8,5))
sns.histplot(df1['price'], bins=30, kde=True)
plt.title("Distribution of Airbnb Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
````

##### 1. Why did you pick the specific chart?

A histogram was chosen because it is the most suitable chart for displaying the distribution of a continuous numerical variable such as price. It helps identify the frequency of listings within different price ranges, making it easy to understand the overall pricing pattern, detect skewness, and identify any unusually high or low-priced listings (outliers).

##### 2. What is/are the insight(s) found from the chart?

The histogram shows that Airbnb listing prices are positively (right) skewed, meaning that most listings are concentrated in the lower price range, approximately between ₹50 and ₹200. As prices increase, the number of listings decreases significantly. Only a few listings are priced above ₹500, with some luxury properties reaching nearly ₹1,700. This indicates that the majority of Airbnb properties are affordable, while premium listings represent only a small portion of the market.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can create a positive business impact. Since most customers prefer affordable accommodations, hosts can set competitive prices within the popular price range to attract more bookings. The platform can also promote budget-friendly listings to increase occupancy rates and customer satisfaction. Additionally, luxury listings can be marketed separately to target premium travelers.

A potential negative insight is that high-priced listings have lower representation, which may indicate lower demand or a limited customer base for expensive properties. If hosts consistently set prices much higher than the market average without offering additional value or amenities, they may experience fewer bookings, reducing occupancy rates and overall revenue. Therefore, pricing strategies should align with customer demand and market trends to maximize business performance.

#### Chart - 9 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
plt.figure(figsize=(10,7))

corr = df2[['price',
           'minimum_nights',
           'number_of_reviews',
           'reviews_per_month',
           'availability_365',
           'number_of_reviews_ltm',
           'calculated_host_listings_count']].corr()

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title("Correlation Heatmap")
plt.show()
````

##### 1. Why did you pick the specific chart?

A heatmap was chosen because it is the most effective visualization for showing the correlation between multiple numerical variables in a single chart. It helps identify the strength and direction of relationships using color intensity and correlation coefficients, making it easier to understand which variables are strongly related and which have little or no relationship.

##### 2. What is/are the insight(s) found from the chart?

The heatmap shows that reviews_per_month and number_of_reviews_ltm have the strongest positive correlation (0.87), indicating that listings receiving more monthly reviews also tend to receive more reviews in the last 12 months. There is also a strong positive correlation (0.68) between number_of_reviews and reviews_per_month, suggesting that popular listings continue to attract reviews over time. On the other hand, price has very weak correlations with most variables, indicating that pricing is not strongly influenced by factors such as availability, number of reviews, or host listing count. Additionally, minimum_nights has a moderate negative correlation (-0.29) with reviews_per_month, implying that listings requiring longer minimum stays tend to receive fewer monthly reviews.
